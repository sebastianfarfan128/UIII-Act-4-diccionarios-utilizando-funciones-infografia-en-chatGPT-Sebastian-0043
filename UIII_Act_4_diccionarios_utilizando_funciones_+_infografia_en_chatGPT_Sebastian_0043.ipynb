## 18. Diseñador

```python
# --- INICIALIZACIÓN ---
# El diccionario principal empieza vacío [6]
Diccionario_diseñador = {}

# --- FUNCIONES DEL SISTEMA [4] ---

def Listar_diseñador():
    """Muestra los datos en forma de tabla bien distribuida [2, 3]."""
    if len(Diccionario_diseñador) == 0:
        print("\n[IA]: El diccionario está vacío actualmente.")
    else:
        # Cabecera de la tabla con alineación profesional [3, 7]
        print("\n" + "="*145)
        print(f"{'NOMBRE (KEY)':<20} | {'PORTAFOLIO':<20} | {'SOFTWARE':<20} | {'ESTILO':<15} | {'CLIENTES':<10} | {'TARIFA':<10} | {'HARDWARE':<15}")
        print("-" * 145)

        # Uso de for clave, valor (items) para recorrer el diccionario [2, 8]
        for nombre, datos in Diccionario_diseñador.items():
            print(f"{nombre:<20} | {datos['portafolio']:<20} | {datos['software']:<20} | {datos['estilo']:<15} | {datos['clientes']:<10} | ${datos['tarifa']:<9} | {datos['hardware']:<15}")
        print("="*145)

def agregar_diseñador():
    """Captura datos de forma interactiva y los guarda en el diccionario [9, 10]."""
    print("\n>>> REGISTRO DE NUEVO DISEÑADOR")
    nombre = input("Nombre del Diseñador (Clave única): ")

    # Se crea un diccionario interno para los valores [1]
    detalles = {
        "portafolio": input("URL Portafolio digital: "),
        "software": input("Software de diseño: "),
        "estilo": input("Estilo artístico: "),
        "clientes": input("Número de clientes activos: "),
        "tarifa": input("Tarifa por proyecto (USD): "),
        "hardware": input("Tipo de hardware: ")
    }

    # Asignación de la entrada al diccionario principal [10, 11]
    Diccionario_diseñador[nombre] = detalles
    print(f"\n[IA]: '{nombre}' ha sido agregado.")
    Listar_diseñador() # Siempre muestra la opción 2 al terminar

def buscar_diseñador():
    """Localiza un diseñador por su clave [10, 12]."""
    nombre = input("\nIngrese el nombre del diseñador a buscar: ")
    if nombre in Diccionario_diseñador: # Verifica si la clave existe [12]
        datos = Diccionario_diseñador[nombre]
        print(f"\n>>> RESULTADO PARA '{nombre}':")
        print(f"Software: {datos['software']} | Estilo: {datos['estilo']} | Tarifa: ${datos['tarifa']}")
    else:
        print("\n[IA]: El diseñador no existe en los registros.")
    Listar_diseñador()

def borrar_diseñador():
    """Elimina un registro del diccionario [13]."""
    nombre = input("\nNombre del diseñador a borrar: ")
    if nombre in Diccionario_diseñador:
        del Diccionario_diseñador[nombre] # Elimina la clave y sus valores [13]
        print(f"\n[IA]: Registro de '{nombre}' eliminado.")
    else:
        print("\n[IA]: Error, el nombre no fue localizado.")
    Listar_diseñador()

def actualizar_diseñador():
    """Modifica los valores de una clave existente [10]."""
    nombre = input("\nNombre del diseñador a actualizar: ")
    if nombre in Diccionario_diseñador:
        print("Ingrese los nuevos datos:")
        Diccionario_diseñador[nombre]["clientes"] = input("Nuevos clientes activos: ")
        Diccionario_diseñador[nombre]["tarifa"] = input("Nueva tarifa: ")
        print(f"\n[IA]: Datos de '{nombre}' actualizados.")
    else:
        print("\n[IA]: No se puede actualizar, el nombre no existe.")
    Listar_diseñador()

# --- BUCLE PRINCIPAL (MENÚ INTERACTIVO) [14, 15] ---

while True:
    print("\n--- MENÚ ADMINISTRADOR 'DISEÑO-PRO' ---")
    print("1. Agregar diseñador")
    print("2. Listar diccionarios de diseñador")
    print("3. Buscar diseñador en diccionario")
    print("4. Borrar diseñador en diccionario")
    print("5. Actualizar diseñador en diccionario")
    print("6. Salir")

    opcion = input("\nSeleccione una instrucción (1-6): ")

    # Llamadas a funciones mediante if-elif [5, 16]
    if opcion == "1":
        agregar_diseñador()
    elif opcion == "2":
        Listar_diseñador()
    elif opcion == "3":
        buscar_diseñador()
    elif opcion == "4":
        borrar_diseñador()
    elif opcion == "5":
        actualizar_diseñador()
    elif opcion == "6":
        print("\n[IA]: Cerrando sistema de administración. Adiós.")
        break # Finaliza el bucle while True [14]
    else:
        print("\n[IA ERROR]: Opción no válida.")
```
